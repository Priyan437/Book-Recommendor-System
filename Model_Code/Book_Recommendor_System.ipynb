**Popularity Based Recommendor System.**

```python
Ratings_with_name = Ratings.merge(Books, on = 'ISBN')
```

```python
num_rating_df = Ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_rating'},inplace = True)
num_rating_df.head()
```

```python
avg_rating_df = Ratings_with_name.groupby('Book-Title')['Book-Rating'].sum().reset_index()
avg_rating_df.rename(columns={'Book-Rating':'sum_rating'},inplace = True)
avg_rating_df['avg_rating'] = avg_rating_df['sum_rating']/num_rating_df['num_rating']
avg_rating_df.head()
```

```python
popular_df = avg_rating_df.merge(num_rating_df,on='Book-Title')
popular_df.head()
```

```python
popular_df = popular_df[popular_df['num_rating']>=250].sort_values('avg_rating',ascending = False).head(50)
popular_df = popular_df.merge(Books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-M','num_rating','avg_rating']]
popular_df.head()
```

## Collaborative Filter based Predictor


```python
Ratings_with_name
```

```python
x = Ratings_with_name.groupby('User-ID').count()['Book-Rating'] > 200
user_with_200_rating_done = x[x].index
```

```python
Filtered_Ratings  = Ratings_with_name[Ratings_with_name['User-ID'].isin(user_with_200_rating_done)]
Filtered_Ratings.shape
```

```python
y = Filtered_Ratings.groupby('Book-Title').count()['Book-Rating'] >= 50
famous_books = y[y].index
```

```python
final_rating = Filtered_Ratings[Filtered_Ratings['Book-Title'].isin(famous_books)]
```

```python
final_rating
```

```python
pt = final_rating.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.head()
```

```python
pt.shape
```

```python
pt.fillna(0,inplace = True)
```

```python
pt.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_score = cosine_similarity(pt)
```

```python
similarity_score.shape
```

```python
def recommend_book(book_name):

  ##We will first fetch the Index of Book
  index = np.where(pt.index==book_name)[0][0]
  similar_items = sorted(list(enumerate(similarity_score[index])), key = lambda x:x[1],reverse = True)[1:6]

  data = []
  for i in similar_items:
    item = []
    temp_df = Books[Books['Book-Title'] == pt.index[i[0]]]
    item.extend(list( temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
    item.extend(list( temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
    item.extend(list( temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values))

    data.append(item)

  return data
```

```python
recommend_book('The Notebook')
```
